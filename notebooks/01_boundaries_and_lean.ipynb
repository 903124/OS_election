# Counties, districts, and how each one leans

First notebook in a series on doing actual things with the
[OS_election](https://github.com/903124/OS_election) datasets —
Wikipedia-sourced election results, tidied into CSVs by the pipeline in that
repo.

The agenda here is deliberately narrow:

1. get real boundary geometry for counties and congressional districts,
2. attach election numbers to the shapes,
3. eyeball a random sample of each — 16 counties, 16 districts — before ever
   drawing all ~3,100 shapes at once.

| thing | comes from | vintage |
|---|---|---|
| presidential results by county | `OS_election/data/presidential/` | 2024 |
| district lean (PVI-style) | `OS_election/data/district_lean/` | 2024 |
| county polygons | Esri mirror of the Census cartographic files | ~1:20m |
| 119th Congress polygons | Esri `USA_119th_Congressional_Districts` | current map, incl. 2024 redraws |

Two honest notes before the code. census.gov 403-blocks this machine, so the
boundaries come from Esri's feature-service mirror of the same cartographic
files, requested with a ~500 m simplification tolerance — the same order of
accuracy as the Census 1:20m originals, and the response shrinks about tenfold.
And "lean" gets a precise definition in section 2; spoiler: it's a two-party
margin relative to the nation, Cook PVI style. It is not a forecast of
anything.


```python
import json
import re
import urllib.parse
import urllib.request
from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import TwoSlopeNorm

REPO = Path("../OS_election")   # cloned from github.com/903124/OS_election
GEO = Path("data/geo")          # cached boundary files
SEED = 14                       # any int — change it, rerun, resample

pd.options.display.float_format = "{:.2f}".format
plt.rcParams.update({
    "figure.dpi": 100, "axes.spines.top": False, "axes.spines.right": False,
    "axes.titlesize": 11,
})

print("geopandas", gpd.__version__, "| pandas", pd.__version__)
```

## 1 · Geometry

Counties and districts both arrive as GeoJSON in EPSG:4326 (lon/lat) and get
reprojected to **EPSG:5070** (Albers equal-area) before plotting — the sane
choice for coloring US shapes, and the same projection the repo's own
crosswalk uses for its area math.

The fetcher is boring on purpose: check the cache, download what's missing,
report counts. `maxAllowableOffset` is the server-side simplification knob;
it takes the 1:20m-ish cartographic files and returns them smaller and just
as wrong in exactly the ways that don't matter at panel scale.


```python
ESRI = "https://services.arcgis.com/P3ePLMYs2RVChkJx/ArcGIS/rest/services"


def esri_geojson(service, page=2000, max_pages=10):
    """Page a Living Atlas feature service into a FeatureCollection dict."""
    url = f"{ESRI}/{service}/FeatureServer/0/query"
    feats, offset = [], 0
    for _ in range(max_pages):
        q = urllib.parse.urlencode(dict(
            where="1=1", outFields="*", outSR=4326, f="geojson",
            resultRecordCount=page, resultOffset=offset,
            maxAllowableOffset=0.005, geometryPrecision=5))
        req = urllib.request.Request(
            f"{url}?{q}", headers={"User-Agent": "lean-notebook/0.1"})
        with urllib.request.urlopen(req, timeout=180) as r:
            d = json.load(r)
        feats += d.get("features", [])
        more = d.get("exceededTransferLimit",
                     d.get("properties", {}).get("exceededTransferLimit"))
        if not more or not d.get("features"):
            break
        offset += page
    return {"type": "FeatureCollection", "features": feats}


def cached(fname, service):
    dest = GEO / fname
    if not dest.exists():
        dest.write_text(json.dumps(esri_geojson(service)))
    return dest


GEO.mkdir(parents=True, exist_ok=True)
districts = gpd.read_file(cached("cd119.geojson",
                                 "USA_119th_Congressional_Districts"))
counties = gpd.read_file(cached("counties_generalized.geojson",
                                "USA_Counties_Generalized_Boundaries"))

print(f"{len(districts)} district polygons, {len(counties)} county polygons")
districts.drop(columns="geometry").head(3)
```

```python
TERRITORIES = {"AS", "GU", "MP", "PR", "VI"}   # geometry ships them, wiki doesn't

fig, ax = plt.subplots(figsize=(7.5, 4.6))
states = districts[~districts["STATE_ABBR"].isin(TERRITORIES)]
states.to_crs(5070).plot(ax=ax, facecolor="none", edgecolor="0.55",
                         linewidth=0.3)
ax.set_axis_off()
ax.set_title(f"all {len(states)} districts at once — the hairball sampling "
             "exists to avoid");
```

## 2 · A lean for every county

The presidential CSV is candidate-level: one row per (county, candidate).
Three steps to a lean:

1. pivot votes wide — D, R, everything else
2. two-party D share = D / (D + R); third parties are summed and set aside,
   exactly as the pipeline's own lean does
3. lean = county share − national share, in points. Positive = bluer than
   the country, negative = redder.

One wrinkle the pipeline already handles and this notebook has to repeat:
Minnesota and North Dakota Democrats run under state-party labels
(Democratic-Farmer-Labor, Democratic-NPL). They count as Democratic.

The national baseline is computed from the county sums themselves, then
checked against the number the lean pipeline used (49.25 for 2024). If those
two ever disagree badly, something upstream broke — that's what the check is
for.


```python
pres = pd.read_csv(REPO / "data/presidential/presidential_results_2024.csv")

DEM = {"democratic", "democratic-farmer-labor", "democratic-npl"}

def party_class(p):
    p = str(p or "").strip().lower()
    return "d" if p in DEM else "r" if p == "republican" else "o"

pres["pc"] = pres["party"].map(party_class)

# DC reports eight wards; fold them into the single county-equivalent
pres.loc[pres["state_code"] == "DC", "county"] = "District of Columbia"

votes = (pres.pivot_table(
            index=["state_code", "county", "subdivision_type"],
            columns="pc", values="votes", aggfunc="sum", fill_value=0)
            .reset_index())
votes["two_party"] = votes["d"] + votes["r"]
votes["d_share"] = 100 * votes["d"] / votes["two_party"]

national = 100 * votes["d"].sum() / votes["two_party"].sum()
lean = pd.read_csv(REPO / "data/district_lean/district_lean_2024.csv")
national_pipeline = lean["national_d_share_two_party_pct"].iloc[0]

print(f"national D two-party share — from county sums : {national:.2f}")
print(f"national D two-party share — pipeline's number : {national_pipeline:.2f}")

votes["lean"] = votes["d_share"] - national_pipeline
votes["lean_label"] = votes["lean"].map(
    lambda v: f"D+{v:.1f}" if v >= 0 else f"R+{-v:.1f}")

print(f"\n{len(votes)} county-equivalents")
votes["lean"].describe()
```

### Matching names to polygons

The wiki tables identify counties by name; the geometry layer carries
NAMELSAD-style names ("Autauga County", "Fairfax city"). The matcher is a
small pile of boring rules, tried in order per row: the raw name, the
suffixed form, the census spelling of particles ("De Witt" vs "DeWitt"),
city-vs-county candidates for Virginia, which names its 133 localities
inconsistently. Everything still unmatched at the end gets printed, not
silently dropped.


```python
SUFFIX_RE = re.compile(r"\s+(COUNTY|PARISH|BOROUGH|CENSUS AREA|CITY AND "
                       r"BOROUGH|CITY|MUNICIPALITY)$")
PARTICLE_RE = re.compile(r"\b(DE|DU|LA|LE|MC) ")


def norm(s):
    """Fold name spellings: no case, no apostrophes/okina, punctuation out."""
    for ch in "ʻ'’":
        s = s.replace(ch, "")
    s = re.sub(r"[^A-Za-z0-9 ]+", " ", s)
    return re.sub(r"\s+", " ", s).upper().strip()


def candidates(row):
    """Wiki name -> ordered list of geometry-name candidates."""
    n, t = str(row["county"]).strip(), row["subdivision_type"]
    if t == "Locality":                    # VA counties + independent cities
        stem = re.sub(r"\s+City$", "", n, flags=re.I)
        keys = [n, f"{stem} County", f"{stem} city", f"{n} County"]
    elif t == "Parish":
        keys = [n, f"{n} Parish"]
    elif t == "County" and not SUFFIX_RE.search(norm(n)):
        keys = [n, f"{n} County"]
    else:                                  # Alaska ships full forms
        keys = [n]
    return list(dict.fromkeys(keys))


counties = counties[counties["STATE_ABBR"] != "PR"]    # no wiki data for PR

lookup = {}
for i, r in counties.iterrows():
    k = norm(r["NAME"])
    lookup[(r["STATE_ABBR"], k)] = i
    lookup[(r["STATE_ABBR"], PARTICLE_RE.sub(r"\1", k))] = i  # De Witt = DeWitt

def match(row):
    for cand in candidates(row):
        k = (row["state_code"], norm(cand))
        if k in lookup:
            return lookup[k]
    return None

votes["geo_idx"] = votes.apply(match, axis=1)
matched = votes["geo_idx"].notna().sum()
print(f"matched {matched}/{len(votes)} county-equivalents "
      f"({100 * matched / len(votes):.1f}%)")
votes.loc[votes["geo_idx"].isna(), ["state_code", "county"]]
```

```python
county_geo = (votes.loc[votes["geo_idx"].notna()]
              .set_index("geo_idx")
              .join(counties[["NAME", "SQMI", "geometry"]], how="left")
              .reset_index())
county_geo = gpd.GeoDataFrame(county_geo, geometry="geometry",
                              crs=counties.crs)

LIM = np.ceil(county_geo["lean"].abs().max())          # symmetric color limits
norm_c = TwoSlopeNorm(vmin=-LIM, vcenter=0, vmax=LIM)
cmap = plt.colormaps["RdBu"]   # blue toward the D end, red toward the R end

print(f"{len(county_geo)} counties with geometry, lean range "
      f"[{county_geo['lean'].min():.1f}, {county_geo['lean'].max():.1f}]")
county_geo[["state_code", "county", "NAME", "lean", "lean_label"]].head(4)
```

## 3 · Sixteen random counties, drawn to shape

Why sample instead of choroplething all ~3,100? Because a full county map is
3,100 overlapping edges dominated by whichever places happen to be biggest —
information-rich, shape-poor. The sample view does the opposite: a handful of
places, each actually visible.

16 counties, uniform at random, seeded so reruns reproduce. Each panel fits
its own shape, so relative size is carried by the printed area, and the
full-distribution histogram below keeps the sample honest. Change `SEED` in
the setup cell and rerun to draw a fresh hand.

Don't be surprised if a draw comes back heavily red — this one did. Uniform
sampling of counties is close to sampling land, and most counties lean
Republican; the median county sits ~20 points right of the nation (see the
`describe()` above). The districts grid in section 4 comes back balanced by
construction, which is exactly the contrast section 5 plots.


```python
def lean_multiples(gdf, n, title_fmt, suptitle):
    """n sampled shapes in a grid, colored by lean on the shared scale."""
    sample = gdf.sample(n, random_state=SEED).to_crs(5070)
    cols = 4
    rows = int(np.ceil(len(sample) / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(3 * cols, 2.7 * rows))
    for ax, (_, row) in zip(axes.flat, sample.iterrows()):
        gpd.GeoSeries([row.geometry], crs=sample.crs).plot(
            ax=ax, color=cmap(norm_c(row["lean"])), edgecolor="0.25",
            linewidth=0.4)
        ax.set_axis_off()
        ax.set_title(title_fmt(row), fontsize=9)
    for ax in axes.flat[len(sample):]:
        ax.set_visible(False)
    sm = plt.cm.ScalarMappable(norm=norm_c, cmap=cmap)
    fig.colorbar(sm, ax=axes, orientation="horizontal", fraction=0.045,
                 pad=0.03, label="lean (points vs national two-party share)")
    fig.suptitle(suptitle, fontsize=12)
    return sample


sample_c = lean_multiples(
    county_geo, 16,
    title_fmt=lambda r: (f"{r['NAME']}, {r['state_code']}\n"
                         f"{r['lean_label']}  ·  {r['SQMI']:.0f} sq mi"),
    suptitle=f"16 counties at random (seed {SEED})");
```

```python
fig, ax = plt.subplots(figsize=(8.5, 3))
ax.hist(county_geo["lean"], bins=80, color="0.88", edgecolor="0.78")
for v in sample_c["lean"]:
    ax.axvline(v, color=cmap(norm_c(v)), lw=2)
ax.axvline(0, color="0.25", lw=0.8, ls="--")
ax.set_xlabel("county lean, 2024 (points)")
ax.set_ylabel("counties")
ax.set_title("all matched counties — sampled sixteen marked in color");
```

## 4 · Same drill, congressional districts

No math this time — the pipeline ships the lean ready-made (`lean_pct`,
`lean_label`; same definition, same national baseline). The work is the
join: the geometry layer ids districts as state-FIPS + CD number, the CSV
says `AL-01`, and at-large seats are coded `00` (or `98` for DC) on the
geometry side but `AL` in the CSV. A two-line mapper handles it.


```python
AT_LARGE = {"AK", "DE", "ND", "SD", "VT", "WY"}

def district_id(props):
    st, cdn = props["STATE_ABBR"], str(props["CDFIPS"])
    if st == "DC" or st in AT_LARGE:      # 98 = DC, 00 = at-large states
        return f"{st}-AL"
    return f"{st}-{cdn.zfill(2)}"

districts["district"] = districts.apply(district_id, axis=1)

cd_geo = districts.merge(
    lean[["district", "lean_pct", "lean_label"]], on="district", how="left")

print(f"{cd_geo['lean_pct'].notna().sum()} of {len(cd_geo)} polygons carry "
      "lean data")
print("no data:",
      sorted(cd_geo.loc[cd_geo["lean_pct"].isna(), "district"].tolist()))
```

```python
cd_matched = cd_geo[cd_geo["lean_pct"].notna()].copy()
cd_matched["lean"] = cd_matched["lean_pct"]

sample_d = lean_multiples(
    cd_matched, 16,
    title_fmt=lambda r: f"{r['district']}\n{r['lean_label']}",
    suptitle=f"16 districts at random (seed {SEED})");
```

```python
bins = np.linspace(-LIM, LIM, 60)
fig, ax = plt.subplots(figsize=(8.5, 3.2))
ax.hist(county_geo["lean"], bins=bins, color="0.85", edgecolor="0.8",
        label=f"counties (n={len(county_geo)})")
ax.hist(cd_matched["lean"], bins=bins, color="#4477aa", alpha=0.45,
        edgecolor="white", label=f"districts (n={len(cd_matched)})")
ax.axvline(0, color="0.2", lw=0.8, ls="--")
ax.set_xlabel("lean (points vs national two-party share)")
ax.set_ylabel("shapes")
ax.legend()
print(f"median |lean| — counties: {county_geo['lean'].abs().median():.1f}, "
      f"districts: {cd_matched['lean'].abs().median():.1f}");
```

## 5 · One state, both layers

Counties nest under congressional districts — every district is a union
of whole or partial counties. Zooming into one state makes the nesting
visible: counties on the left, the congressional districts built from
them on the right, both colored on the same lean scale.

Colorado (CO) is a clean example: 64 counties, 8 congressional districts,
rectangular shape. The eastern-plains and western-slope counties are
deep red; the Denver–Boulder–Front Range corridor is blue, and that
corridor is exactly what the 1st and 2nd districts trace. The at-large
states from section 4 don't appear here — they only have one district,
so the two panels would just be the same shape.


```python
STATE = "CO"

state_counties = county_geo[county_geo["state_code"] == STATE].to_crs(5070)
state_cds = (cd_matched[cd_matched["STATE_ABBR"] == STATE]
             .sort_values("district").to_crs(5070))

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
state_counties.plot(ax=axes[0], column="lean", cmap=cmap, norm=norm_c,
                    edgecolor="0.4", linewidth=0.3)
axes[0].set_title(f"{STATE} — {len(state_counties)} counties by lean",
                   fontsize=10)
axes[0].set_axis_off()

state_cds.plot(ax=axes[1], column="lean", cmap=cmap, norm=norm_c,
               edgecolor="0.2", linewidth=0.8)
for _, r in state_cds.iterrows():
    c = r.geometry.representative_point()
    axes[1].text(c.x, c.y, f"{r['district']}\n{r['lean_label']}",
                 ha="center", va="center", fontsize=8)
axes[1].set_title(f"{STATE} — {len(state_cds)} congressional districts by lean",
                   fontsize=10)
axes[1].set_axis_off()

sm = plt.cm.ScalarMappable(norm=norm_c, cmap=cmap)
fig.colorbar(sm, ax=axes, orientation="horizontal", fraction=0.045,
             pad=0.04, label="lean (points vs national two-party share)")
fig.suptitle(f"{STATE}: counties and congressional districts, both colored by lean",
             fontsize=12)

state_cds[["district", "lean_label", "lean"]].sort_values("lean", ascending=False)
```

## Takeaways

The notebook above does four things in order:

1. **Pull boundary geometry.** 441 congressional-district polygons and
   3,144 county polygons come from Esri's mirror of the Census
   cartographic files (EPSG:4326 → 5070 for plotting, cached under
   `data/geo/`). `maxAllowableOffset=0.005` shrinks the response ~10×
   with no loss visible at panel scale.
2. **Compute a Cook-PVI-style lean for every county-equivalent.**
   Pivot the 2024 presidential CSV wide (D / R / other), take D / (D + R),
   subtract the national two-party share. The county sums reproduce the
   pipeline's 49.25 baseline exactly — the cross-check passes. 3,143
   county-equivalents; median lean R+20.4, range R+45.7 to D+44.1.
3. **Match names to polygons.** A small pile of boring rules (raw name,
   suffixed form, particles, VA city-vs-county) lands 3,135 of 3,143
   county-equivalents (99.7%). The 8 misses are all Connecticut — the
   geometry layer ships them as towns, the wiki export lists them as
   counties.
4. **Sample 16 of each, don't choropleth all 3,100.** A uniform county
   sample ≈ sampling land, so this draw came back heavily red; the
   districts grid is balanced by construction. The comparison histogram
   makes the contrast explicit — median |lean| is 21.3 for counties
   vs 10.0 for districts. Change `SEED` in the setup cell to draw a fresh
   hand.
5. **Zoom one state.** Colorado (64 counties, 8 CDs) shows both layers
   on the same lean scale: counties on the left, the districts built
   from them on the right. The Front Range corridor (CO-01, CO-02) sits
   at D+18 to D+30 while the eastern-plains and western-slope counties
   drive CO-03, CO-04, CO-05 to R+1 to R+4 — the nesting is exactly
   what the section 4 contrast predicts at the national scale.

**Caveats that actually change the picture.** The Esri simplification is
fine at panel scale but loses small features; the CT name miss is a
known wiki-vs-Census naming mismatch (8 of 3,143, not silently dropped);
and the territories (AS, GU, MP, PR, VI) ship geometry but no wiki
election data, so they're excluded from the joins throughout.
